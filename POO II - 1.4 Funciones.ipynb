## Cierre (5 min)

### Lo que aprendimos hoy
| Concepto | Java | Python |
|---|---|---|
| Definir | `static double f(double x) { … }` | `def f(x):` |
| Devolver | `return x;` | `return x` |
| Sin valor de retorno | `void` | devuelve `None` |
| Opcionales | sobrecarga de métodos | valores por defecto `def f(x, y=1)` |
| Por nombre | no existe | `f(y=2, x=1)` |
| Varios valores | clase o arreglo | `return a, b` (tupla) |
| Varargs | `double... xs` | `*args` |
| Documentar | Javadoc `/** … */` | docstring `""" … """` |
| Lambda | `x -> x * 2` | `lambda x: x * 2` |
